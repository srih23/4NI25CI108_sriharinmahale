```python
grade_info=pd.DataFrame({
    'grade':['A','B','C','D','F'],
    'grade_description' : [
        'Excellent',
        'Good',
        'Average',
        'Below Average',
        'Fail'
    ]
})
print("\ngrade information dataframe:")
print(grade_info)
```

```python
merged=pd.merge(
    df,
    grade_info,
    on='grade',
    how='left'
)
print("\nMerged Dataframe:")
print(merged.head())
```

```python
grade_details = pd.DataFrame({
    'description': [
        'Excellent',
        'Good',
        'Average',
        'Below average',
        'Fail'
    ],
    'performance_level': [
        'Very High',
        'High',
        'Low',
        'Very Low',
        'Fail'
    ]
}, index=['A', 'B', 'C', 'D', 'F'])

print(grade_details.head())

grade_counts = df['grade'].value_counts().to_frame('student_count')
print(grade_counts.head())

joined = grade_counts.join(grade_details)
print(joined)
```

```python
#checking missing values 
print(df.isnull().sum())
```

```python
#removing missing values 
df.dropna()
```

```python
#checking duplicated rows
print(df.duplicated().sum())
```

```python
#removing duplicated rows 
df = df.drop_duplicates()
```

```python
#check duplicates again
print(df.duplicated().sum())
```

```python
#display statistical summary 
print(df.describe())
```

```python
#set graph style
sns.set_style("whitegrid")
```

```python
#create graph area
plt.figure(figsize=(8,5))
```

```python
#create histogram
sns.histplot(data=df , x="SepalLengthCm", bins=10)
#add title
plt.title("Distribuiton of Sepal Length")
#add x-axis label 
plt.xlabel("Sepal length (cm)")
#add y-label label 
plt.ylabel("Number of flowers")
#display
plt.show()
```

```python
#create bos plot 
sns.boxplot(data=df  , x="Species" , y="SepalLengthCm")
#add title
plt.title("Sepal length by Species")
#add x-axis label 
plt.xlabel("Species")
#add y-axis label
plt.ylabel("Sepal Length (cm)")
#display graph
plt.show()
```

```python
#create graph area 
plt.figure(figsize=(8,5))
#create scatter plot
sns.scatterplot(
    data=df,
    x="SepalLengthCm",
    y="PetalLengthCm",
    hue="Species"
)
#add title
plt.title("Sepal length vs Petal length")
#add x-label
plt.xlabel("Petal length (cm)")
#add y-label
plt.ylabel("Sepal Length (cm)")
#dislay
plt.show()
```

```python
#create pair plot 
sns.pairplot(df, hue="Species")
#display graph 
plt.show()
```

```python
#Select numerical columns
numeric_columns = [
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]
#calculate correlation 
correlation = df[numeric_columns].corr()
#isplay correlation matrix
print(correlation)
```

```python
#heat map
#create graph area 
plt.figure(figsize=(8,6))
#create correlation heat map 
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
#add title
plt.title("Correlation heatmap")
#display graph
plt.show()
```

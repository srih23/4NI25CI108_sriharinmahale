```python
#creating copy of dataset
df_mean = df.copy()
```

```python
#calculate the mean of BuildingArea
mean_building_area=df_mean["BuildingArea"].mean()

#replacing missing value with mean
df_mean["BuildingArea"]=df_mean["BuildingArea"].fillna(mean_building_area)
print("\nMean building Area : ",mean_building_area)
print("missing BuildingArea values after Mean Imputation: ")
print(df_mean["BuildingArea"].isnull().sum())
```

```python
#creating copy of dataset
df_median=df.copy()
```

```python
#calculating median of YearBuilt
median_year_built = df_mean["YearBuilt"].median()

#replacing missing value with median
df_median["YearBuilt"]=df_median["YearBuilt"].fillna(median_year_built)
print("\nMedian year built : ",median_year_built)
print("missing yearbuilt values after Median Imputation: ")
print(df_median["YearBuilt"].isnull().sum())
```

```python
#creating copy for forward fill imputation
df_ffill = df.copy()
```

```python
df_ffill["Car"] = df_ffill["Car"].ffill()
```

```python
print("\nMissing car values after forward fill\n")
print(df_ffill["Car"].isnull().sum())
```

```python
comparison=pd.DataFrame({
    "Original BuildingArea" : df["BuildingArea"].head(15) , 
    "Mean imputed BuiltArea" : df_mean["BuildingArea"].head(15) , 
    "Original yearbuilt" : df["YearBuilt"].head(15) , 
    "Median imputed yearbuilt" : df_median["YearBuilt"].head(15),
    "Original Car : " : df["Car"].head(15) , 
    "Forward Filled Car" : df_ffill["Car"].head(15)
})

print("\nComparison of imputation methods")
print(comparison)
```

```python
#create copy of dataset
df_processed = df.copy()
```

```python
#most frequent value 
council_mode = df_processed["CouncilArea"].mode()[0]

#replacing missing values with mode
df_processed["CouncilArea"] = df_processed["CouncilArea"].fillna(council_mode)
print("\nCouncil Mode : ",council_mode)
print("\nMissing Categories values after imputation : ")
print(df_processed[["CouncilArea"]].isnull().sum())
```

```python
df_encoded=pd.get_dummies(
    df_processed , 
    columns=["Type" , "Method"], 
    dtype=int
)
print("\nDataset After One-Hot Encoding : ")
print(df_encoded.head())
```

```python
df_label=df.copy()
label_encoder = LabelEncoder()

df_label["CouncilArea_Encoded"] = label_encoder.fit_transform(
    df_label["CouncilArea"]
)

print("\nLabel Encoding of CouncilArea:")
print(
    df_label[["CouncilArea", "CouncilArea_Encoded"]].head(10)
)
```

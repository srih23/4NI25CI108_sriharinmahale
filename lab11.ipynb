```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
print("jajajaja")
```

```python
df=pd.read_csv(r"C:\Users\Srihari\Downloads\Mall_Customers.csv")
print(df.shape)
print(df.head())
print(df.describe())
print("missing values",df.isnull().sum().sum())
```

```python
features =["Annual Income (k$)","Spending Score (1-100)"]
x=df[features]
scaler=StandardScaler()
x_scaled=scaler.fit_transform(x)
print("y")
```

```python
k_values=range(2,11)
inertias=[]
silhouettes=[]
for k in k_values:
    km=KMeans(n_clusters=k,n_init=10,random_state=42)
    km.fit(x_scaled)
    inertias.append(km.inertia_)
    silhouettes.append(silhouette_score(x_scaled,km.labels_))
    print(f"K={k:2d} inertia = {km.inertia_:8.2f} silhouette={silhouettes[-1]:.3f}")
fig,ax=plt.subplots(1,2,figsize =(12,4))    
ax[0].plot(list(k_values),inertias,marker = "o")
ax[0].set_xlabel("no of clusters (k)")
ax[0].set_ylabel("inertia(wccs)")
ax[0].set_title("elbow method (bend)")
ax[1].plot(list(k_values),silhouettes,marker = "o",color="green")
ax[1].set_xlabel("no of clusters (K)")
ax[1].set_ylabel("silhouette score")
ax[1].set_title("silhouette score (high)")
plt.tight_layout
plt.show()
```

```python
best_k =5
kmeans=KMeans(n_clusters = best_k,n_init=10,random_state = 42)
df["Cluster"]=kmeans.fit_predict(x_scaled)
print("customers per cluster :\n",df["Cluster"].value_counts().sort_index())
print("final silhouette score :",round(silhouette_score(x_scaled,df["Cluster"]),3))
```

```python
centroids=scaler.inverse_transform(kmeans.cluster_centers_)
plt.figure(figsize = (8,6))
sns.scatterplot(
    data=df,x=features[0],y=features[1],hue="Cluster",palette ="tab10",s=70
)
plt.scatter(centroids[:,0],centroids[:,1],marker="X",s=250,c="black",label="Centroids")
plt.title(f"k-means(k={best_k})")
plt.legend()
plt.tight_layout()
plt.show()
```
